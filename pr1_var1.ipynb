```python
import pandas as pd
import numpy as np

# Таблица продаж
sales = pd.DataFrame({
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
                'Laptop', 'Tablet', 'Phone', 'Laptop', 'Tablet'],
    'Quantity': [1, 2, 1, 3, 1, 2, 1, 4, 1, 2],
    'Price': [50000, 30000, 20000, 50000, np.nan,
              50000, 20000, 30000, 50000, np.nan],
    'Region': ['Moscow', 'SPb', 'Kazan', 'Moscow', 'SPb',
               'Kazan', np.nan, 'SPb', 'Moscow', 'Kazan']
})

# Таблица товаров
products = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor'],
    'Category': ['Electronics', 'Electronics',
                 'Electronics', 'Accessories'],
    'Weight_kg': [2.5, 0.2, 0.5, 3.0]
})

print("Таблица продаж:")
display(sales)

print("Таблица товаров:")
display(products)
```

```python
sales_by_product = sales.groupby('Product')['Quantity'].sum()

print("Общее количество продаж для каждого продукта:")
print(sales_by_product)
```

```python
avg_price_region = sales.groupby('Region')['Price'].mean()

print("Средняя цена для каждого региона:")
print(avg_price_region)
```

```python
mean_price = sales['Price'].mean()

sales['Price'] = sales['Price'].fillna(mean_price)

print("Таблица после заполнения пропусков:")
display(sales)
```

```python
sales_sorted = sales.sort_values(
    by='Quantity',
    ascending=False
)

print("Отсортированная таблица:")
display(sales_sorted)
```

```python
sales_products = pd.merge(
    sales,
    products,
    on='Product',
    how='inner'
)

print("Объединённая таблица:")
display(sales_products)
```
